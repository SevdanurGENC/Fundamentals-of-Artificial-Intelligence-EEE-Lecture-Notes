## 4. Koşullar: if / elif / else

Girinti (4 boşluk) bloğu belirler — Python'da süslü parantez yoktur.

```python
akim = 18.2   # A

if akim > 16:
    print("Aşırı akım! Röle açılıyor.")
elif akim > 12:
    print("Uyarı: yük yüksek.")
else:
    print("Normal çalışma.")
```

## 5. Döngüler: for ve while

```python
olcumler = [14.1, 15.8, 17.3, 12.0, 16.5]
for i, akim in enumerate(olcumler):          # enumerate: indeks + değer
    durum = "AŞIRI" if akim > 16 else "normal"
    print(f"Örnek {i}: {akim} A -> {durum}")

print("---")
for k in range(5):                            # 0,1,2,3,4
    print(k, k ** 2)
```

```python
# while: RC devresinde kondansatör deşarjı, gerilim %5'e inene kadar
V0, R, C, dt = 5.0, 1000, 1e-3, 0.001   # V, ohm, F, s (dt küçük -> daha doğru)
V, t = V0, 0.0
while V > 0.05 * V0:
    V = V - (V / (R * C)) * dt           # dV/dt = -V/(RC)  (Euler adımı)
    t += dt
print(f"%5'e inme süresi ≈ {t:.3f} s   (teorik 3RC = {3*R*C:.3f} s)")
```

**Deneyin:** `dt` değerini 0.05 yapın. Sonuç teorik değerden ne kadar uzaklaşıyor? (Sayısal integrasyonda adım büyüklüğünün etkisi.)

## 6. Fonksiyonlar

```python
def seri(*direncler):
    """Seri bağlı dirençlerin eşdeğeri: toplam."""
    return sum(direncler)

def paralel(*direncler):
    """Paralel bağlı dirençlerin eşdeğeri."""
    return 1 / sum(1 / R for R in direncler)

def gerilim_bolucu(vin, r1, r2):
    """Vout = Vin * R2 / (R1 + R2)"""
    return vin * r2 / (r1 + r2)

print(seri(100, 220, 470))                 # 790
print(paralel(100, 100))                   # 50.0
print(round(paralel(100, 220, 470), 2))    # 58.68
print(gerilim_bolucu(12, 10_000, 4_700))   # 3.84
```

### Örnek 2.1 — Direnç renk kodu

```python
RENK = {"siyah": 0, "kahverengi": 1, "kırmızı": 2, "turuncu": 3, "sarı": 4,
        "yeşil": 5, "mavi": 6, "mor": 7, "gri": 8, "beyaz": 9}

def direnc_degeri(b1, b2, b3):
    return (10 * RENK[b1] + RENK[b2]) * 10 ** RENK[b3]

print(direnc_degeri("kahverengi", "siyah", "kırmızı"))   # 1000
print(direnc_degeri("sarı", "mor", "turuncu"))            # 47000
```

## 7. NumPy: diziler ve vektörel işlemler

```python
import numpy as np

a = np.array([2, 5, 1, 7])                 # vektör
M = np.array([[1, 0, 2], [4, 3, 5]])       # matris
print(a.shape, M.shape)                    # (4,) (2, 3)
print(M[1, 2])                             # 5
print(M[:, 0])                             # 1. sütun -> [1 4]
```

```python
# Ohm kanunu — döngü yok, işlem her elemana uygulanır
I = np.array([0.5, 1.0, 1.5, 2.0])   # A
R = 100
V = I * R
P = V * I
print("V:", V)
print("P:", P)
print("ortalama:", V.mean(), " std:", V.std(), " max:", V.max(), " argmax:", V.argmax())
```

## 8. Sinyal üretme ve ilk grafik (matplotlib)

```python
import matplotlib.pyplot as plt

fs = 1000                                # örnekleme frekansı (Hz)
t = np.arange(0, 0.1, 1/fs)              # 0..0.1 s
v = 230*np.sqrt(2)*np.sin(2*np.pi*50*t)  # ideal 50 Hz sinüs
rng = np.random.default_rng(0)
v_olcum = v + rng.normal(0, 25, size=t.size)   # gürültülü ölçüm

print("İdeal RMS :", np.sqrt(np.mean(v**2)).round(2))
print("Ölçüm RMS :", np.sqrt(np.mean(v_olcum**2)).round(2))

plt.figure(figsize=(8, 3))
plt.plot(t*1000, v_olcum, label="ölçüm", lw=1)
plt.plot(t*1000, v, label="ideal", lw=2)
plt.xlabel("Zaman (ms)"); plt.ylabel("Gerilim (V)"); plt.legend(); plt.grid(alpha=.3)
plt.show()
```

**Deneyin:** Gürültü standart sapmasını (25) 0, 10, 50 yapıp ölçüm RMS'ini gözlemleyin. Neden ideal değerden hep *büyük*?

## 9. Hız farkı: döngü mü, vektör mü? (Örnek 2.3)

```python
import time
x = np.arange(1_000_000, dtype=float)

t0 = time.time()
kare_liste = [xi**2 for xi in x]      # Python döngüsü
t1 = time.time()
kare_np = x**2                        # NumPy vektörel
t2 = time.time()
print(f"Döngü: {t1-t0:.3f} s   NumPy: {t2-t1:.4f} s   hızlanma ≈ {(t1-t0)/max(t2-t1,1e-9):.0f}x")
```
