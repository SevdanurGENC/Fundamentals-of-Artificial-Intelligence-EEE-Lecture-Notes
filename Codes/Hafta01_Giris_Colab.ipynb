## 3. Değişkenler: değerlere isim vermek

```python
gerilim = 230        # volt
akim = 2.5           # amper
guc = gerilim * akim # watt
print("Güç =", guc, "W")

# Değişkenin türünü öğrenmek
print(type(gerilim), type(akim), type(guc))
```

## 7. Örnek 1.2: en yakın komşu ile tahmin

Derste elle hesapladık; şimdi bilgisayar hesaplasın. (NumPy'ı gelecek hafta ayrıntılı öğreneceğiz; şimdilik satırları okuyup ne yaptığını tahmin etmeye çalışın.)

```python
import numpy as np

X = np.array([[24, 9], [31, 18], [20, 3], [28, 14]])   # her satır: sıcaklık, saat
y = np.array([118, 206, 42, 122])                       # tüketim (kW)
yeni = np.array([26, 15])                               # tahmin etmek istediğimiz gün

uzakliklar = np.sqrt(((X - yeni) ** 2).sum(axis=1))    # Öklid uzaklığı, her satır için
print("Uzaklıklar:", uzakliklar.round(2))
en_yakin = uzakliklar.argmin()                          # en küçük uzaklığın indeksi
print("En yakın örnek:", "ABCD"[en_yakin], "-> tahmin:", y[en_yakin], "kW")
```

### Gelecek haftaya hazırlık
Aşağıdaki hücreyi çalıştırıp her satırın ne yaptığını tahmin edin. Gelecek hafta listeleri ve döngüleri ayrıntılı işleyeceğiz.

```python
olcumler = [229.8, 231.2, 230.5, 228.9, 230.1]   # bir liste
for v in olcumler:                                # her eleman için tekrar
    if v > 230:
        print(v, "-> yüksek")
    else:
        print(v, "-> normal")
```
